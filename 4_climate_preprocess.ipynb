# Preprocessing climate data and landcover

```python
import glob
import remote_sensing_processor as rsp
```

```python
rasters = glob.glob(r'CHELSA**.tif')
```

## Data matching

First we match the climatology and landcover rasters to the same extent and resoulution as Landsat images.

```python
for raster in rasters:
    in_file = raster
    out_file = raster.split("_")[1] + "_unp.tif"
    clip = r'G:/Work/hmao-landsat-processed/clipper.gpkg'
    ref = r'G:/Work/hmao-landsat-processed/1984/summer/hmao-B1-1984.tif'
    rsp.process(in_file, out_file, clip=clip, reference_raster=ref)
```

```python
clip = r'G:/Work/hmao-landsat-processed/clipper.gpkg'
ref = r'G:/Work/hmao-landsat-processed/1984/summer/hmao-B1-1984.tif'
rsp.process("Landcover.tif", "Landcover_unp.tif", clip=clip, reference_raster=ref, fill_nodata=True)
```

## Normalization

Then we normalize the data.

For `bio5`, `bio6` and `gsl` we use Dynamic World normalization.

```python
rsp.get_normalization_params.dynamic_world('bio5_unp.tif', percentiles=[30, 70], nodata=0)
```

```python
rsp.normalize.dynamic_world('bio5_unp.tif', percentile1=7.985144138336182, percentile2=7.988203525543213, output_path='bio5_p.tif', nodata=0)
```

```python
rsp.get_normalization_params.dynamic_world('bio6_unp.tif', percentiles=[30, 70], nodata=0)
```

```python
rsp.normalize.dynamic_world('bio6_unp.tif', percentile1=7.8220438957214355, percentile2=7.829232692718506, output_path='bio6_p.tif', nodata=0)
```

```python
rsp.get_normalization_params.dynamic_world('gsl_unp.tif', percentiles=[30, 70])
```

```python
rsp.normalize.dynamic_world('gsl_unp.tif', percentile1=4.795790672302246, percentile2=4.927253723144531, output_path='gsl_p.tif')
```

For `gsp` we use min-max normalization

```python
rsp.get_normalization_params.percentile('gsp_unp.tif', percentiles=[1, 99])
```

```python
rsp.normalize.min_max('gsp_unp.tif', minimum=(2690 * 0.9), maximum=(3755 * 1.1), output_path='gsp_p.tif', clip_values=True)
```

For landcover we use min-max normalization.

```python
rsp.get_normalization_params.min_max('Landcover_unp.tif')
```

```python
rsp.normalize.min_max('Landcover_unp.tif', minimum=0, maximum=100, output_path='Landcover_p.tif')
```
