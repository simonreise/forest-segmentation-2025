# Generating datasets

Here we generate ML-ready datasets from our data

```python
import glob
import pathlib
import remote_sensing_processor as rsp
```

```python
path_x = glob.glob(r"G:/Work/hmao-landsat-processed/2024/**/*.json") + glob.glob(r"G:/Work/hmao-index/2024/**/*.json") + glob.glob(r"G:/Work/hmao-dem/*-p.json") + glob.glob(r"G:/Work/hmao-climate/*_p.json")
path_x
```

```python
path_y = r"G:/Work/hmao-forestry/hmao_forestry_values_filtered_denorm.gpkg"
```

## Generating semantic segmentation dataset

As semantic segmentation and regression datasets have different structure (e.g. different dtype) of target variables, we have to create these datasets separately.

Semantic segmentation dataset is split to train, val and test subdatasets in 3-1-1 proportion, samples are shuffled.

```python
rsp.semantic.generate_tiles(
    x=path_x, 
    y=[
        {'name': 'species', 'path': path_y, 'burn_value': 'Species'}, 
        {'name': 'bonity', 'path': path_y, 'burn_value': 'Bonity'},
        {'name': 'type', 'path': path_y, 'burn_value': 'Type'}
    ],
    output="dataset_semantic.rspds",
    shuffle=True,
    split={'train': 3, 'val': 1, 'test': 1},
    tile_size=256,
)
```

## Generating regression dataset

It is generated the same way as semantic segmentation dataset.

```python
rsp.regression.generate_tiles(
    x=path_x, 
    y=[
        {'name': 'age', 'path': path_y, 'burn_value': 'Age'}, 
        {'name': 'volume', 'path': path_y, 'burn_value': 'Volume'},
    ],
    output="dataset_regression.rspds",
    shuffle=True,
    split={'train': 3, 'val': 1, 'test': 1},
    tile_size=256,
    y_nodata=-1,
)
```

## Generating map dataset

This dataset contains 2024 data, but is for mapping only, not for training. So it does not need target variable (and can be used with both semantic segmentation and regression models), not split to subdatasets and not shuffled.

```python
rsp.semantic.generate_tiles(
    x=path_x, 
    y=None,
    output="dataset_map.rspds",
    shuffle=False,
    tile_size=256,
)
```
