### Fires

Here we filter out the areas that were exposed to fires in last 40 years, but have age more than 40 years according to forestry data.

We got info about fires by calculating dNBR from our landsat data and removing the noize.

```python
fires = gpd.read_file("fires_mask.gpkg")
```

```python
file["idx"] = file.index
```

```python
file['orig_area'] = file.to_crs(32642).geometry.area
```

```python
intersections = gpd.overlay(file, fires.to_crs(file.crs), how='intersection')
```

```python
intersections['intersect_area'] = intersections.to_crs(32642).geometry.area
```

```python
intersections['overlap_ratio'] = intersections['intersect_area'] / intersections['orig_area']
```

```python
result = intersections[(intersections['overlap_ratio'] >= 0.5) & (intersections['Amz1'] > 40)]
```

```python
file = file.loc[~file["idx"].isin(result["idx"])]
```

```python
file = file.drop(columns=['idx', 'orig_area'])
```

```python
import rioxarray as rxr
from geocube.api.core import make_geocube

with rxr.open_rasterio(r"G:\Work\hmao-index\2024\summer\ndvi.tif") as ndvi:
    out_grid = make_geocube(
        vector_data=file.reset_index(),
        measurements=["index"],
        like=ndvi, # ensure the data are on the same grid
    )
    out_grid["ndvi"] = (ndvi.dims, ndvi.values, ndvi.attrs, ndvi.encoding)
    grouped_elevation = out_grid.drop_vars("spatial_ref").groupby(out_grid.index)
    grid_mean = grouped_elevation.mean().rename({"ndvi": "ndvi_s"}).to_dataframe()
    file = file.reset_index().merge(grid_mean, on="index", how="left").drop(columns=['index'])
```

```python
import rioxarray as rxr
from geocube.api.core import make_geocube

with rxr.open_rasterio(r"G:\Work\hmao-index\2024\winter\ndvi.tif") as ndvi:
    out_grid = make_geocube(
        vector_data=file.reset_index(),
        measurements=["index"],
        like=ndvi, # ensure the data are on the same grid
    )
    out_grid["ndvi"] = (ndvi.dims, ndvi.values, ndvi.attrs, ndvi.encoding)
    grouped_elevation = out_grid.drop_vars("spatial_ref").groupby(out_grid.index)
    grid_mean = grouped_elevation.mean().rename({"ndvi": "ndvi_w"}).to_dataframe()
    file = file.reset_index().merge(grid_mean, on="index", how="left").drop(columns=['index'])
```

```python
file["ndvi_w"] = file["ndvi_w"].fillna(0)
```
