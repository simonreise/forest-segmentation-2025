# Calculate dNBR

```python
import glob
import pathlib

import xarray as xr
import rioxarray as rxr
```

```python
nbrs = glob.glob(r'G:/Work/hmao-index/**/**/NBR.tif')
```

```python
years = list(range(1984, 2024))
seasons = ['summer', 'winter']
```

```python
change = None

for season in seasons:
    for year in years:
        nbrfile1 = f'G:/Work/hmao-index/{str(year)}/{season}/NBR.tif'
        nbrfile2 = f'G:/Work/hmao-index/{str(year + 1)}/{season}/NBR.tif'
        
        with rxr.open_rasterio(nbrfile1, chunks=True, lock=True) as nbr1, rxr.open_rasterio(nbrfile2, chunks=True, lock=True) as nbr2:
            if change is None:
                change = xr.zeros_like(nbr1, dtype='uint16')
                change = change.rio.write_nodata(0)
            dnbr = nbr1 - nbr2
            threshold = 0.3
            burnt = dnbr > threshold
            change = change.where(burnt==0, year).persist()
        print(str(year))
    
    writer = change.rio.to_raster(
        f'G:/Work/hmao-index/change_{season}.tif',
        compress="deflate",
        PREDICTOR=2,
        ZLEVEL=9,
        BIGTIFF="IF_SAFER",
        tiled=True,
        NUM_THREADS="ALL_CPUS",
        compute=True,
        lock=True,
    )
```
