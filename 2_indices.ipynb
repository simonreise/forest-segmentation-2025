# Calculating vegetation indices

Here we calculate NDVI and NBR from our Landsat data

```python
import glob
import pathlib
import re
import remote_sensing_processor as rsp
```

```python
landsats = glob.glob(r'G:/Work/hmao-landsat/**/**/')
```

```python
for landsat in landsats:
    in_file = landsat
    out_file = landsat.replace('hmao-landsat', 'hmao-index')
    year = re.findall(r'\d\d\d\d', in_file)[0]
    if not pathlib.Path(out_file).exists():
        rsp.calculate_index('NDVI', in_file, out_file, bands = {'N': f'hmao-B4-{year}', 'R': f'hmao-B3-{year}'}, nodata=-9999)
        rsp.calculate_index('NBR', in_file, out_file, bands = {'N': f'hmao-B4-{year}', 'S2': f'hmao-B7-{year}'}, nodata=-9999)
        print(out_file)
```
