# DEM preprocessing

Here we preprocess FABDEM data. We calculate slope, curvature and aspect, normalize the DEM, and then clip all the data.

```python
import glob
import pathlib
import remote_sensing_processor as rsp
```

```python
rsp.dem.slope('dem-unp.tif', 'slope-unp.tif', normalize=True, nodata=-9999)
```

```python
rsp.dem.curvature('dem-unp.tif', 'curvature-unp.tif', normalize=True, nodata=-9999)
```

```python
rsp.dem.aspect('dem-unp.tif', 'aspect-unp.tif', normalize=True, nodata=-9999)
```

```python
rsp.get_normalization_params.dynamic_world('dem-unp.tif', percentiles=[30, 70], nodata=-9999)
```

```python
rsp.normalize.dynamic_world('dem-unp.tif', percentile1=4.04305126783455, percentile2=4.553876891600541, output_path='dem-norm-unp.tif', nodata=-9999)
```
