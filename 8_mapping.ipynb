# Mapping

In this notebook we map the model predictions for every year to get yearly maps of forest parameters we model.

```python
import glob
import pathlib
import remote_sensing_processor as rsp
```

```python
year = "1984"
```

## Generating the dataset

We generate a dataset to use it for mapping for a specific year. It does not need target variable, splitting or shuffling.

```python
path_x = glob.glob(f"G:/Work/hmao-landsat-processed/{year}/**/*.json") + glob.glob(f"G:/Work/hmao-index/{year}/**/*.json") + glob.glob(r"G:/Work/hmao-dem/*-p.json") + glob.glob(r"G:/Work/hmao-climate/*_p.json")
path_x
```

```python
rsp.semantic.generate_tiles(
    x=path_x,
    y=None,
    output=f"dataset_map_{year}.rspds",
    shuffle=False,
    tile_size=256,
)
```

## Mapping dominant tree species

```python
train_dataset = {"path": "G:/Work/hmao-train-data/dataset_semantic.rspds", "sub": "train", "y": "species"}
map_dataset = {"path": f"dataset_map_{year}.rspds", "sub": "all"}
```

```python
rsp.semantic.generate_map(
    map_dataset, 
    model=r'G:/Work/hmao-train-data/models/segformer_species.ckpt', 
    output=f"{year}/map_species_{year}.tif", 
    reference_dataset=train_dataset,
    batch_size=16,
)
```
